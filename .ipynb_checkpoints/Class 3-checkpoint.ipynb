# Linked Lists
In general, there are two varieties of *linked lists*:
1. Singly linked lists (or uni-directional lists)
2. Doubly linked lists (or bi-directional lists)
   

```python
mylist = []

help(mylist)

for number in range(11):
    mylist.append(number)

print(mylist)
# print(mylist[2])

```

```python
class SinglyLinkedList:
    class Node:
        def __init__(self, datum):
            self.datum = datum
            self.next = None

    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, datum):
        new_node = self.Node(datum)
        if not self.head:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node

    def insert(self, index, datum):
        new_node = self.Node(datum)
        if index == 0:
            new_node.next = self.head
            self.head = new_node
            if not self.tail:
                self.tail = new_node
        else:
            prev = None
            current = self.head
            count = 0
            while current and count < index:
                prev = current
                current = current.next
                count += 1
            if count == index:
                prev.next = new_node
                new_node.next = current
                if not current:
                    self.tail = new_node
            else:
                raise IndexError("Index out of range")

    def remove(self, value):
        if self.head:
            current = self.head
            prev = None
            while current:
                if current.datum == value:
                    if prev:
                        prev.next = current.next
                    else:
                        self.head = current.next
                    if not current.next:
                        self.tail = prev
                    return
                prev = current
                current = current.next
            raise ValueError("Value not found")
        else:
            raise IndexError("List empty")

    def index(self, value):
        count = 0
        current = self.head
        while current:
            if current.datum == value:
                return count
            current = current.next
            count += 1
        raise ValueError("Value not found")

    def __len__(self):
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count

    def __str__(self):
        if not self.head:
            return "[]"
        out = "["
        current = self.head
        if current:
            out += f"{current.datum}"
        while current.next:
            current = current.next
            out += f", {current.datum}"
        out += "]"
        return out

    def __getitem__(self, index):
        count = 0
        current = self.head
        while current:
            if count == index:
                return current.datum
            current = current.next
            count += 1
        raise IndexError("Index out of range")

# Example usage
mylist = SinglyLinkedList()
print(mylist)

mylist.append(1)
print(mylist)

mylist.append(2)
print(mylist)
```
